# Customer Data Cleaning

## Objective

Clean and validate customer master data before loading into the analytics database.

## Cleaning Tasks

- Identify missing values
- Remove duplicate customer IDs
- Standardize customer channels
- Correct regional inconsistencies
- Validate customer records


```python
import pandas as pd
from pathlib import Path
```

```python
raw_path = Path("../data/raw/customers.csv")

clean_path = Path("../data/cleaned/customers_clean.csv")
```

```python
customers = pd.read_csv(raw_path)

customers.head()
```

```python
customers.shape
```

```python
customers.info()
```

```python
customers.isnull().sum()
```

```python
customers.duplicated().sum()
```

```python
customers["customer_id"].duplicated().sum()
```

```python
customers[
    customers["customer_id"].duplicated(keep=False)
]
```

```python
customers_clean = customers.copy()
```

```python
customers_clean["customer_name"] = (
    customers_clean["customer_name"]
    .fillna("Unknown Customer")
)
```

```python
customers_clean["customer_id"].max()
```

```python
customers_clean.loc[
    customers_clean["customer_id"].duplicated(),
    "customer_id"
] = customers_clean["customer_id"].max() + 1
```

```python
customers_clean["customer_id"].duplicated().sum()
```

```python
customers_clean["channel"].value_counts()
```

```python
customers_clean["region"].value_counts()
```

```python
region_mapping = {
    "Wesst": "West",
    "Northeest": "Northeast"
}

customers_clean["region"] = (
    customers_clean["region"]
    .replace(region_mapping)
)
```

```python
customers_clean["region"].value_counts()
```

```python
customers_clean.info()
```

```python
customers_clean.isnull().sum()
```

```python
customers_clean.to_csv(
    clean_path,
    index=False
)
```

## Summary

The customer dataset was cleaned and validated by:

- Handling missing customer names
- Resolving duplicate customer identifiers
- Validating channel categories
- Correcting regional inconsistencies
- Confirming data completeness

The cleaned dataset is ready for downstream analytics and database loading.
